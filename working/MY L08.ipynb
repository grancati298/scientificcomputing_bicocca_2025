## Q4: So slow

Pick a piece of your code that you dislike. It's sooooo slooooow. 
- Profile it.
- Find the bottleneck.
- Rewrite that bottleneck making a better use of numpy arrays, or jitting it with Numba.
- Enjoy your faster code

```python
import random
import time

def monte_carlo_pi_slooow(n):
    count_inside = 0
    for _ in range(n):
        x = random.random()
        y = random.random()
        if x**2 + y**2 <= 1.0:
            count_inside += 1
    return 4 * count_inside / n

result = monte_carlo_pi_slooow(5000000)
print(f"Result for Pi: {result}")
```

```python
import cProfile
import pstats

with cProfile.Profile() as pr:
    monte_carlo_pi_slooow(5000000)

stats = pstats.Stats(pr)
stats.sort_stats('cumtime').print_stats(5)
```

```python
import numpy as np

def monte_carlo_pi_fast(n=5000000):
    
    points = np.random.rand(n, 2)
    distances = np.sum(points**2, axis=1)
    
    count_inside = np.sum(distances <= 1.0)
    
    return 4 * count_inside / n

result_fast = monte_carlo_pi_fast(5000000)
print(f"Result for Pi: {result_fast}")
```

```python
with cProfile.Profile() as pr:
    monte_carlo_pi_fast(5000000)

stats = pstats.Stats(pr)
stats.sort_stats('cumtime').print_stats(5)
```
