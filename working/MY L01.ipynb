## Q5: Foxes and dogs

### Part 1. Short words

Let's practice functions.  Here's a simple function that takes a string and returns a list of all the 4 letter words:

```python
def four_letter_words(message):
    words = message.split()
    four_letters = [w for w in words if len(w) == 4]
    return four_letters
```

Write a version of this function that takes a second argument, n, that is the word length we want to search for

```python
def find_words_of_length_n(message, n):
    words = message.split()
    answer = [w for w in words if len(w) == n]
    return answer

message = "The quick brown fox jumps over the lazy dog"

print(f"3 letter words in the sentence are:", find_words_of_length_n(message, 3))
print(f"5 letter words in the sentence are:", find_words_of_length_n(message, 5))
```

### Part 2: Panagrams

A _panagram_ is a sentence that includes all 26 letters of the alphabet, e.g., "_The quick brown fox jumps over the lazy dog_."

Write a function that takes as an argument a sentence and returns `True` or `False`, indicating whether the sentence is a panagram.

```python
import string

def find_pangram(phrase):
    alphabet = string.ascii_lowercase
    phrase = phrase.lower() 
    
    for letter in alphabet:
        if letter not in phrase:
            return False
            
    return True

text1 = "The quick brown fox jumps over the lazy dog"
print(find_pangram(text1),f". The phrase '{text1}' is a pangram.")

text2 = "The wizard quickly jinxed the gnomes before they vaporized."
print(find_pangram(text2),f". The phrase '{text2}' is a pangram.")

text3 = "Giulia went to bed since she was tired"
print(find_pangram(text3),f". The phrase '{text3}' is not a pangram.")
```
